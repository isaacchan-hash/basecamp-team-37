# GDP Data Cleaning

We'll remove the letter prefix (A, B, C, etc.) from the start of each column name in the GDP table.


```python
# Remove letter prefix from gdp column names
gdp.columns = gdp.columns.str.replace(r'^[A-Z]\r?\n', '', regex=True)
gdp.head()
```
